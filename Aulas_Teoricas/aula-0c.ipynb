 #### Comparadores e Operadores Lógicos



 Para o computador tomar decisões, precisamos fazer perguntas de "Sim ou Não" (True ou False).



 * **Comparadores Matemáticos:** `>`, `<`, `>=`, `<=`

 * **Igualdade:** `==` (É igual a?) e `!=` (É diferente de?)

 * **`and` (E):** Só é verdadeiro se **TUDO** for verdadeiro.

 * **`or` (Ou):** É verdadeiro se **PELO MENOS UMA** coisa for verdadeira.

 * **`not` (Não):** Inverte o valor (True vira False, False vira True).

 -----------------------------

 #### Tomada de Decisão: If, Elif, Else



 Como fazemos o código escolher caminhos diferentes?



 * **`if` (Se):** A primeira pergunta. Só roda se for True.

 * **`elif` (E se):** Se a pergunta de cima for False, tenta essa. Pode ter quantos quiser.

 * **`else` (Senão):** O caminho final. Roda se todas as perguntas anteriores falharem. Não tem condição associada.

 * **Atenção total à indentação!**

 -------------------

 #### Obs: Valores "Falsos" (Falsy)



 No Python, não precisamos sempre de uma verificação explícita como `== True`.

 Alguns valores vazios ou nulos são automaticamente considerados `False` em um `if`:



 * O número zero: `0`

 * Uma string de texto vazia: `""`

 * O valor nulo: `None`

 * Uma lista vazia: `[]`

 -------------------

 #### Laços de Repetição: For Loop



 Usamos o `for` quando temos uma **coleção** (como uma lista ou string) ou quando sabemos **exatamente** quantas vezes queremos repetir algo.



 * Ele "passa" por cada item, um de cada vez.

 * **A função `range(início, fim)`:** É a melhor amiga do `for` quando queremos apenas contar números em sequência.

 -------------------

 #### Laços de Repetição: While Loop



 O `while` (Enquanto) repete o código de forma contínua **enquanto** uma condição for verdadeira.



 * Excelente para processos de tentativa e erro, ou quando não sabemos quando o laço deve acabar.

 * **Cuidado:** Se a condição nunca virar False, você cria um **Loop Infinito** e trava o programa!

 -------------------

 #### Introdução a Funções (`def`)



 Se você está copiando e colando o mesmo código em vários lugares, está na hora de criar uma função.

 Funções empacotam o código para ser reutilizado (Princípio DRY - Don't Repeat Yourself).



 * **`def`:** A palavra que usamos para definir uma função.

 * **Argumentos:** As informações que passamos para dentro da função trabalhar.

 * **`return`:** O resultado final que a função devolve para o resto do programa.
