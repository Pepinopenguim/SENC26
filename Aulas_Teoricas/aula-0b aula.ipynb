#### Agrupamento de Dados: Coleções

Muitas vezes, precisamos guardar várias informações juntas em uma única variável. 
O Python nos oferece 3 principais formas de fazer isso, cada uma com uma finalidade:


-------------------
#### 1. Listas (`list`)

A estrutura mais comum e flexível. Pense em uma lista de tarefas ou de compras.

* **Sintaxe:** Usam colchetes `[ ]`
* **Mutáveis:** Podemos adicionar, alterar ou remover itens à vontade.
    * .append(), .extend(), .pop(), .remove()
* **Ordenadas:** A ordem de inserção **importa**. O índice começa no **0**.
    * usamos \[colchetes\] para obter itens específicos 
    * podemos mudar sua ordem com .sort() por exemplo
* **Permitem Repetição:** Aceitam valores duplicados.
* Ex: `compras = ["pão", "leite", "pão"]`


```python
numeros = [1, 2, 3, 4, 7, 10, 10, 11, 5, 6]

numeros.sort()

numeros
```



-------------------
#### 2. Tuplas (`tuple`)

São as "irmãs travadas" das listas. Pense em dados que não devem mudar, como os meses do ano ou coordenadas geográficas (latitude e longitude).

* **Sintaxe:** Usam parênteses `( )`
* **Imutáveis:** Uma vez criadas, **NÃO** podem ser alteradas (são mais seguras e ocupam menos memória).
* **Ordenadas:** Também possuem índice (começando no 0).
* **Permitem Repetição:** Aceitam valores duplicados.
* Ex: `cor_branca_rgb = (255, 255, 255)`


```python
numeros = tuple("abcdefg")

```



-------------------
#### 3. Sets (`set`)

Inspirados nos "Conjuntos" da matemática. Úteis quando a exclusividade é o que importa. Pense em uma lista de CPFs cadastrados em um site.

* **Sintaxe:** Usam chaves `{ }` (apenas valores separados por vírgula).
* **SEM Repetição:** Ignoram e excluem itens duplicados automaticamente. Isso é extremamente útil
* **Desordenados:** Não possuem ordem fixa e não têm índice (não dá para buscar a "posição 0").
* **Mutáveis:** Podemos adicionar e remover itens.
    * .add(), .remove()
* São **muito** mais eficientes em obter dados
* Ex: matrículas da UnB, CPFs

```python
matriculas = {2256824354, 199285436}

matriculas.add(1093405932)

matriculas.remove(2256824354)

matriculas
```

```python
"""
Este é um exemplo para explicar como sets são eficientes!
Não precisa entender o código aqui, apenas observar os resultados!
"""

import timeit

uma_lista = list(range(1, 100000))
um_set = set(range(1, 100000))

item = 99999

tempo_lista = timeit.timeit(lambda: item in uma_lista, number=1000)
tempo_set = timeit.timeit(lambda: item in um_set, number=1000)

print(f"Tempo na Lista: {tempo_lista:.6f} segundos")
print(f"Tempo no Set:   {tempo_set:.6f} segundos")
```

 #### Agrupamento de Dados: Dicionários



 Dicionários (`dict`) são perfeitos para criar relações. Eles funcionam com o sistema de **Chave : Valor**.

 É excelente para tabelas de preços, propriedades de materiais, etc.



 * Criados usando chaves `{ }`

 * não confundir com sets!

 * criam a relação `{CHAVE: VALOR}`



```python
geladeira_do_CA = {'Monster': 8.50, 'azedinho': [1,2,3], 2:"asdf"}

geladeira_do_CA['pepsi'] = 3.0

geladeira_do_CA
```
