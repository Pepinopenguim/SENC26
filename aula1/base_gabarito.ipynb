### Desafio 1 - Importação e carregamento dos dados

Os dados criados pelo software sempre estarão no formato `"ano_mes_dia.json"`. Por exemplo: `"2025_02_15.json"`.

Crie uma função chamada `carregar_dados_obra` com argumentos `ano`, `mes` e `dia` que:
1. Construa o nome do arquivo no formato correto
2. Abra e leia o arquivo JSON
3. Retorne o dicionário contendo os dados do projeto. Separe os dados em diferentes variáveis, como por exemplo:

```{python}
segmentos = dados["segmentos"]
params = dados["parametros_tecnicos"]
simulacao = dados["simulacao_ambiental"]
```

**Dicas:**
- Use a biblioteca `json` para converter os dados em um dicionário (veja `json.load()`)
- O número do dia e mês devem possuir sempre 2 dígitos (ex: `"02"`, `"15"`)
- Dentro de uma f-string, podemos preencher dígitos com o seguinte formato: `{valor:0{quantidade}d}`. Exemplo: `{mes:02d}` produz `"02"` se `mes = 2`
- Para abrir um arquivo, usamos `with open(nome_arquivo, "r", encoding="utf-8") as arquivo:`
- Pesquise: "como ler arquivo json em python" ou "f-string padding zeros python"


````python
import json

def carregar_dados_obra(ano, mes, dia):
    
    arq = f"{ano}_{mes:02d}_{dia:02d}.json"
    
    with open(arq, 'r', encoding="utf-8") as arquivo:
        dados = json.load(arquivo)

    return dados


carregar_dados_obra(2025, 4, 23)

dados_carregados = carregar_dados_obra(2025, 4, 23)

print(dados_carregados)

segmentos = dados_carregados["segmentos"]
params = dados_carregados["parametros_tecnicos"]
simulacao = dados_carregados["simulacao_ambiental"]
````

### Desafio 2 - Área de pavimento novo por segmento

Agora que os dados estão carregados, seu chefe quer saber a **área de pavimento novo** que será executada em cada segmento da rodovia. O pavimento novo corresponde apenas à **faixa de alargamento** (diferença entre a largura futura e a largura atual). Esse valor é essencial para orçamento de asfalto e planejamento da equipe de pavimentação.

**Fórmula:** `área_pavimento_novo = (largura_futura - largura_atual) × comprimento`

**Instruções:**
1. Acesse a lista de segmentos dentro do dicionário de dados
2. Para cada segmento, calcule a área de pavimento novo usando a fórmula acima
3. Armazene esses valores em uma lista
4. **Armazene o valor da área no segmento no próprio dicionário**
5. Calcule a área total somando os valores da lista
6. Exiba os resultados formatados

**Dicas:**
- Um loop `for` pode percorrer a lista de segmentos
- Para acessar um valor dentro do dicionário: `segmento["largura_atual_m"]`
- A largura futura é sempre maior que a atual (não haverá valores negativos)
- Para formatar números com separador de milhar: `{valor:,.0f}` (exibe `6.250` ao invés de `6250`)
- Pesquise: "como iterar sobre lista de dicionarios em python" ou "como formatar números com milhar python"


````python
lista_de_areas_pav_novo = []

for segmento in segmentos:
    comp_m = segmento["comprimento_m"]
    larg_atual_m = segmento["largura_atual_m"]
    larg_futura_m = segmento["largura_futura_m"]

    area_pav_novo_m2 = (larg_futura_m - larg_atual_m) * comp_m

    lista_de_areas_pav_novo.append(area_pav_novo_m2)

    segmento["area_pav_novo_m2"] = area_pav_novo_m2

    print(f"Segmento {segmento["id"]}: {area_pav_novo_m2} m2")

print(f"Área total: {sum(lista_de_areas_pav_novo)} m2")
````

### Desafio 3 - Volume de movimentação de terra (condicionais)

Para cada segmento, calcule o volume de terra a ser escavado ou aterrado. Copie o escrito anteriormente! A fórmula é:

**Volume = (largura_futura - largura_atual) × comprimento × profundidade**

A **profundidade** varia conforme o tipo de solo:
- Argila: 0,8 m
- Areia: 0,5 m
- Rocha: 0,3 m

**Instruções:**
1. Percorra todos os segmentos
2. Para cada um, identifique o tipo de solo usando uma estrutura condicional (`if/elif/else`)
3. Atribua a profundidade correspondente
4. Calcule o volume e **armazene no próprio segmento**
5. Some todos os volumes para obter o volume total movimentado

**Dicas:**
- Os parâmetros técnicos (`profundidade_corte_m`) estão dentro do dicionário `dados["parametros_tecnicos"]`
- Você pode acessar diretamente: `profundidade = dados["parametros_tecnicos"]["profundidade_corte_m"][tipo_solo]`
- O operador `-` entre duas larguras pode dar negativo? 



````python
volume_total_m3 = 0

for segmento in segmentos:
    area_pav_novo_m2 = segmento["area_pav_novo_m2"]

    # calculo da altura
    tipo_do_solo = segmento["tipo_solo"].lower()

    if tipo_do_solo == "argila":
        altura_m = 0.8
    elif tipo_do_solo == "areia":
        altura_m = 0.5
    else:
        altura_m = 0.3

    volume_pav_novo_m3 = area_pav_novo_m2 * altura_m

    segmento["volume_pav_novo_m3"] = volume_pav_novo_m3

    volume_total_m3 += volume_pav_novo_m3


volume_total_m3

    
````

### Desafio 4 - Cálculo de caminhões e viagens (math.ceil)

**Dicas:**

### Desafio 5 - Compensação ambiental (random + condicional)

A licença ambiental exige a compensação pelas árvores suprimidas. Para cada segmento, você precisa:

1. Calcular quantas árvores serão cortadas:
   ```
   árvores_cortadas = área_expansão_ha × árvores_por_ha
   ```
   Onde `área_expansão_ha = (largura_futura - largura_atual) × comprimento / 10000`

2. Plantar **2 mudas** para cada árvore cortada

3. **Simular** a taxa de sobrevivência das mudas: um valor aleatório entre 70% e 95%

4. Calcular quantas mudas efetivamente sobreviverão

**Instruções:**
1. Use `random.uniform(0.70, 0.95)` para gerar a taxa de sobrevivência de cada segmento
2. Para resultados reprodutíveis, use `random.seed(42)` no início do código
3. Some os totais de árvores cortadas, mudas plantadas e mudas sobreviventes
4. Exiba os resultados por segmento e os totais finais

**Dicas:**
- Biblioteca `random` precisa ser importada: `import random`
- `random.uniform(a, b)` retorna um float entre `a` e `b`
- Para fixar a semente aleatória: `random.seed(42)` – isso garante que todos os alunos terão os mesmos valores "aleatórios"
- Pesquise: "python random uniform" ou "como gerar numeros aleatorios entre 0 e 1 python"

**Exemplo de saída:**
```
Segmento 1: 45 árvores cortadas | 90 mudas plantadas | 76 sobrevivem (84%)
Segmento 2: 52 árvores cortadas | 104 mudas plantadas | 88 sobrevivem (85%)
...
Total de árvores suprimidas: 320
Total de mudas plantadas: 640
Total de mudas sobreviventes: 537
```


````python
import math
import random

random.seed(42)

taxa_sob_min = dados_carregados["simulacao_ambiental"]["taxa_sobrevivencia_mudas_min"]
taxa_sob_max = dados_carregados["simulacao_ambiental"]["taxa_sobrevivencia_mudas_max"]


for segmento in segmentos:
    area_pav_novo_ha = segmento["area_pav_novo_m2"] / 10000

    num_arvores_cortadas = math.ceil(segmento["arvores_por_ha"] * area_pav_novo_ha)
    
    num_mudas_plantadas = num_arvores_cortadas * 2

    num_arvores_sobreviventes = math.floor(num_mudas_plantadas * random.uniform(taxa_sob_min, taxa_sob_max))

    
    print(f"Segmento {segmento["id"]}: {num_arvores_cortadas} árvores cortadas | {num_mudas_plantadas} mudas plantadas | {num_arvores_sobreviventes} sobrevivem ({num_arvores_sobreviventes / num_mudas_plantadas * 100:02f}%)")

````

### Desafio 6 - Controle de qualidade do asfalto (loop aninhado + lógica condicional)

O controle de qualidade exige que a densidade do asfalto esteja entre **2,30 e 2,50 g/cm³**. Para cada segmento, foram coletadas **4 amostras** em diferentes posições da pista.

**Regras de classificação:**
- **APROVADO:** todas as 4 amostras dentro da especificação
- **CONDICIONAL:** 3 amostras dentro da especificação (requer reteste)
- **REPROVADO:** menos de 3 amostras dentro da especificação

**Instruções:**
1. Percorra todos os segmentos
2. Para cada segmento, percorra suas 4 amostras de densidade
3. Conte quantas amostras estão dentro da faixa especificada
4. Classifique o segmento conforme as regras
5. Exiba o resultado e conte quantos segmentos caíram em cada categoria

**Dicas:**
- Você terá **dois loops**: um para os segmentos, e dentro dele, um para as amostras
- Para verificar se um valor está entre dois números: `if minimo <= valor <= maximo`
- A lista de amostras está em `segmento["densidade_amostras_asfalto_g_cm3"]`
- Você pode contar usando um acumulador (`dentro = 0`) que incrementa a cada amostra válida
- Pesquise: "loop dentro de loop python nested loops" ou "como contar itens que atendem condicao python"

**Exemplo de saída:**
```
Segmento 1: amostras [2.35, 2.42, 2.38, 2.31] → 4/4 OK → APROVADO
Segmento 2: amostras [2.28, 2.35, 2.40, 2.33] → 3/4 OK → CONDICIONAL
Segmento 3: amostras [2.45, 2.48, 2.52, 2.44] → 3/4 OK → CONDICIONAL
Segmento 4: amostras [2.33, 2.38, 2.41, 2.36] → 4/4 OK → APROVADO
Segmento 5: amostras [2.40, 2.38, 2.42, 2.39] → 4/4 OK → APROVADO
Segmento 6: amostras [2.25, 2.30, 2.35, 2.28] → 2/4 OK → REPROVADO
Segmento 7: amostras [2.36, 2.42, 2.39, 2.44] → 4/4 OK → APROVADO
Segmento 8: amostras [2.31, 2.34, 2.38, 2.32] → 4/4 OK → APROVADO

Resumo:
   APROVADOS: 5
   CONDICIONAIS: 2
   REPROVADOS: 1
```

````python
qualidade_asfalto = []

for segmento in segmentos:
    amostras_aprovadas = []
    for amostra in segmento["densidade_amostras_asfalto_g_cm3"]:
        # if amostra >= 2.3 and amostra <= 2.5:
        #     amostras_aprovadas.append(True)
        
        # else:
        #     amostras_aprovadas.append(False)
        amostras_aprovadas.append(amostra >= 2.3 and amostra <= 2.5)

    num_aprovados = 0
    for resultado_amostra in amostras_aprovadas:
        if resultado_amostra:
            num_aprovados += 1

    if num_aprovados >= 4:
        resultado_segmento = "Aprovado".upper()
    elif num_aprovados == 3:
        resultado_segmento = "Condicional".upper()
    else:
        resultado_segmento = "Reprovado".upper()

    qualidade_asfalto.append(resultado_segmento)

    print(f"Segmento {segmento["id"]}: {resultado_segmento}")


print(f"""======
Resumo
    APROVADO: {len([i for i in qualidade_asfalto if i == "APROVADO"])}
    CONDICIONAL: {len([i for i in qualidade_asfalto if i == "CONDICIONAL"])}
    REPROVADO: {len([i for i in qualidade_asfalto if i == "REPROVADO"])}
""")
````
