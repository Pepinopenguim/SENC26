 ### Parte 1: Ingestão e Limpeza de Dados



 **Objetivo:** Carregar arquivos CSV e realizar o tratamento de tipagens de dados e valores nulos.



 Os arquivos disponibilizados (`{CODIGO_ESTACAO}_Vazoes.csv` e `{CODIGO_ESTACAO}_ResumoDescarga.csv`) possuem cabeçalhos de metadados, utilizam ponto e vírgula (`;`) como delimitador e vírgula (`,`) como separador decimal.



 **Dicas:**

 - Verifique as linhas iniciais do arquivo em um editor de texto puro para identificar a partir de qual linha os dados tabulares começam. Utilize o argumento `skiprows` na função `read_csv()`.

 - Para lidar com o padrão decimal brasileiro em múltiplas colunas simultaneamente, a aplicação de uma função anônima (`lambda`) em conjunto com os métodos `.map()` ou `.apply()` é altamente recomendada. Ela deve substituir a vírgula por ponto (`.replace(',', '.')`) antes da conversão explícita para o tipo `float`.

 - O método `.dropna(subset=[...])` é essencial para eliminar registros corrompidos ou não medidos que interfeririam nos cálculos posteriores.

```python
# GABARITO #

CODIGO_ESTACAO = 46543000

# 1. Carregamento e limpeza do histórico de vazões
vazoes = pd.read_csv(f"data//{CODIGO_ESTACAO}_Vazoes.csv", sep=";", skiprows=13, encoding="latin-1", index_col=False)
vazoes = vazoes.dropna(subset=['Media'])

# 2. Carregamento e limpeza da curva de descarga (Cota x Vazão)
descarga = pd.read_csv(f"data//{CODIGO_ESTACAO}_ResumoDescarga.csv", sep=";", skiprows=13, encoding="latin-1", index_col=False)

# Limpeza e conversão do padrão decimal brasileiro no DataFrame de vazões
for col in ["Media", "Maxima", "Minima"]:
    vazoes[col] = vazoes[col].map(lambda valor: float(str(valor).replace(",", ".")))

# Limpeza e conversão do padrão decimal brasileiro no DataFrame de descargas
for col in ["Cota", "Vazao"]:
    descarga[col] = descarga[col].astype(str).str.replace(",", ".").astype(float)

descarga = descarga.dropna(subset=['Cota', 'Vazao'])

display(descarga.head(3))
vazoes.info()

```

 ### Parte 2: Agregação Temporal de Múltiplas Variáveis



 **Objetivo:** Manipular tipos de datas e agrupar dados temporais para extrair estatísticas consolidadas.



 Os dados originais de vazão estão em base diária e mensal. Para a análise do comportamento em macroescala, é necessário calcular as vazões máximas, médias e mínimas para cada ano.



 **Dicas:**

 - A conversão da coluna de data de string para o tipo datetime exige o uso da função `pd.to_datetime()`, especificando o formato original (`format='%d/%m/%Y'`).

 - Com a coluna no formato correto, acesse o ano diretamente através da propriedade `.dt.year`.

 - O método `.agg()` acoplado ao `.groupby()` permite calcular múltiplas estatísticas simultaneamente. Você pode nomear a nova coluna e especificar a função de agregação matemática correspondente (ex: `Nome_Nova_Coluna='max'`).

```python
# GABARITO #

# Conversão da coluna de texto para formato datetime
vazoes['Data'] = pd.to_datetime(vazoes['Data'], format='%d/%m/%Y')

# Extração do ano
vazoes['Ano'] = vazoes['Data'].dt.year

# Agrupamento para cálculo das vazões anuais
vazoes_anuais = vazoes.groupby('Ano')['Media'].agg(
    Media_Max='max',
    Media_Media='mean',
    Media_Min='min'
).reset_index()

display(vazoes_anuais.head())

```

 ### Parte 3: Regressão Polinomial da Curva de Descarga



 **Objetivo:** Estabelecer uma relação matemática empírica (modelo) entre Cota e Vazão utilizando a biblioteca Numpy.



 Na hidrologia, a Curva-Chave (Cota x Vazão) frequentemente segue um comportamento que pode ser aproximado por uma função polinomial ou de potência. Criaremos um modelo polinomial ajustado aos dados observados em campo.



 **Dicas:**

 - Importe a biblioteca `numpy`.

 - Utilize a função `np.polyfit(x, y, grau)` para encontrar os coeficientes da equação de melhor ajuste. Um polinômio de grau 2 é um bom ponto de partida prático para curvas de rios estáveis.

 - Passe os coeficientes obtidos para a classe `np.poly1d()`. Isso gerará um objeto que atua como uma função matemática (`f(x)`).

 - Para plotar a linha de tendência de forma contínua e suave, crie um vetor contendo valores de Vazão (`x`) uniformemente espaçados entre o mínimo e o máximo medidos. A função `np.linspace(valor_min, valor_max, numero_de_pontos)` é ideal para este fim.

 - Aplique a função criada no passo anterior sobre este vetor de pontos para obter as Cotas correspondentes (`y`).

```python
# GABARITO #
import numpy as np

# 1. Definição do grau e cálculo dos coeficientes
grau_polinomio = 2
coeficientes = np.polyfit(descarga['Vazao'], descarga['Cota'], grau_polinomio)

# 2. Construção da função matemática avaliável
funcao_tendencia = np.poly1d(coeficientes)

# 3. Geração dos vetores para a linha de tendência no gráfico
vazao_tendencia = np.linspace(descarga['Vazao'].min(), descarga['Vazao'].max(), 100)
cota_tendencia = funcao_tendencia(vazao_tendencia)

print(f"Equação do Polinômio (Grau {grau_polinomio}):\n", funcao_tendencia)

```

 ### Parte 4: Visualização Interativa de Dados



 **Objetivo:** Gerar gráficos interativos consolidados e renderizá-los no navegador web.



 É necessário apresentar as séries temporais das múltiplas estatísticas anuais em um único gráfico, e incorporar o modelo matemático calculado na Parte 3 ao gráfico de dispersão das medições de campo.



 **Dicas:**

 - A função `px.line()` é capaz de plotar múltiplas linhas simultaneamente se você fornecer uma lista com os nomes das colunas ao parâmetro `y` (ex: `y=['Coluna_1', 'Coluna_2']`).

 - Para sobrepor a curva da regressão polinomial ao gráfico de dispersão original gerado pelo `px.scatter()`, utilize o método `.add_scatter()` na figura gerada, especificando `x` e `y` com os vetores criados na Parte 3, e ajustando o parâmetro `mode='lines'`.

 - A exportação em formato `.html` com o parâmetro `auto_open=True` garante a execução limpa e interativa no navegador padrão.

```python
# GABARITO #
import plotly.express as px

# Gráfico 1: Série Histórica de Vazão Diária (Contínua)
fig_vazoes = px.line(
    vazoes,
    x='Data',
    y='Media',
    title=f'Série Histórica de Vazão Diária - Estação {CODIGO_ESTACAO}',
    labels={'Media': 'Vazão (m³/s)', 'Data': 'Data do Registro'}
)

# Gráfico 2: Séries Anuais Consolidadas (Max, Med, Min)
fig_vazoes_anuais = px.line(
    vazoes_anuais,
    x="Ano",
    y=["Media_Max", "Media_Media", "Media_Min"],
    title=f"Estatísticas de Vazões Anuais - Estação {CODIGO_ESTACAO}",
    labels={"value": 'Vazão (m³/s)', 'variable': "Estatística"}
)

# Gráfico 3: Curva de Descarga - Dispersão e Linha de Regressão
fig_descarga = px.scatter(
    descarga,
    x='Vazao',
    y='Cota',
    title=f'Curva de Descarga (Cota x Vazão) com Regressão - Estação {CODIGO_ESTACAO}',
    labels={'Vazao': 'Vazão (m³/s)', 'Cota': 'Cota (cm)'}
)

# Adição da linha de tendência calculada com Numpy
fig_descarga.add_scatter(
    x=vazao_tendencia,
    y=cota_tendencia,
    mode='lines',
    name='Modelo Polinomial',
    line=dict(color='black', dash='dash')
)

# Renderização dos resultados finais no navegador
fig_vazoes_anuais.write_html("vazoes_anuais.html", auto_open=True)
fig_descarga.write_html("descarga.html", auto_open=True)
```
